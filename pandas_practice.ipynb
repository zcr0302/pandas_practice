```python
import numpy as np
import pandas as pd

df = pd.read_csv('data/data.csv',encoding='latin-1')
```

```python
df.info()
df.head(5)
```

```python
df.isna().sum()
```

change id to str type, date to date type

```python
df['CustomerID'] = df['CustomerID'].astype('Int64')
df['CustomerID'] = df['CustomerID'].astype('string')
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])
```

```python
df['CustomerID'].nunique()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
```

save returned InvoiceNo as a single df

```python
returns_df = df[df['InvoiceNo'].str[0] == 'C']
df = df[df['InvoiceNo'].str[0] != 'C']
```

```python
neg_qty = df[df['Quantity'] <= 0]
neg_price = df[df['UnitPrice'] <= 0]
print(len(neg_qty))
print(len(neg_price))
```

```python
neg_price.head(10)
```

```python
neg_qty.head(10)
```

3853 rows of error price/quantity in total, most of them have unitprice = 0 and no customerID, decide to delete.

```python
df = df[(df['Quantity'] > 0) & (df['UnitPrice'] > 0)]
```

```python
df['CustomerID'].isna().sum()
```

a very large amount of number, use fillna().

```python
df[df['CustomerID'] == '00000']
```

```python
df['CustomerID'] = df['CustomerID'].fillna('00000')
df[df['CustomerID'] == '00000']
```

```python
df['Description'] = df['Description'].str.strip().str.upper()
```

```python
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
```

```python
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day'] = df['InvoiceDate'].dt.day
df['Weekday'] = df['InvoiceDate'].dt.weekday
df['Hour'] = df['InvoiceDate'].dt.hour
df['Minute'] = df['InvoiceDate'].dt.minute
```

```python
df['Y-M'] = df['InvoiceDate'].dt.to_period('M')
```

```python
df.head(5)
```

```python
df = df.set_index(df['InvoiceDate']).sort_index()
```

```python
df.loc['2011-3']
```

```python
daily_sale = df.resample('D')['TotalPrice'].sum()
weekly_sale = df.resample('W')['TotalPrice'].sum()
monthly_sale = df.resample('MS')['TotalPrice'].sum()
print(daily_sale.head(5))
print(weekly_sale.head(5))
print(monthly_sale.head(5))
```

```python
customers = df[df['CustomerID'] != '00000']
customers.head(5)
```

```python
FirstBuy = customers.groupby('CustomerID')['InvoiceDate'].min()
LastBuy = customers.groupby('CustomerID')['InvoiceDate'].max()
```

```python
CountryStats = df.groupby('Country').agg(
    orders = ('InvoiceNo','nunique'),
    customers = ('CustomerID','nunique'),
    sales = ('TotalPrice','sum')
).sort_values('sales', ascending=False)
CountryStats.head(5)
```

```python
print(df.groupby('InvoiceNo').agg({
    'Quantity':'sum',
    'TotalPrice':'sum'
}).sort_values('Quantity', ascending=False).head(10))

print(df.groupby('InvoiceNo').agg({
    'Quantity':'sum',
    'TotalPrice':'sum'
}).sort_values('TotalPrice', ascending=False).head(10))
```

```python
print(df.groupby('Weekday')['InvoiceNo'].count().sort_values(ascending=False))

print(df.groupby('Hour')['InvoiceNo'].count().sort_values(ascending=False))
```

pivot table

```python
monthly_country = df.pivot_table(
    index='Y-M',
    columns='Country',
    values='TotalPrice',
    aggfunc='sum'
)
monthly_country
```

```python
total_sales = df['TotalPrice'].sum()
avg_order_value = total_sales / df['InvoiceNo'].nunique()
print(avg_order_value)
```

```python
monthly = df.groupby('Y-M')['TotalPrice'].sum().to_frame('Sales')
monthly['MoM'] = monthly['Sales'].pct_change()
print(monthly)
```

use transform to get price ratio

```python
df['invoice_price'] = df.groupby('InvoiceNo')['TotalPrice'].transform('sum')
df['ratio'] = df['TotalPrice'] / df['invoice_price']
df['ratio']
```

```python
sold = df.groupby('StockCode')['Quantity'].sum().reset_index(name='sold_qty')
returned = returns_df.groupby('StockCode')['Quantity'].sum().abs().reset_index(name='returned_qty')
sold_and_returned = pd.merge(sold, returned, on='StockCode', how='left')
sold_and_returned['returned_qty'] = sold_and_returned['returned_qty'].fillna(0)
sold_and_returned['return_rate'] = sold_and_returned['returned_qty'] / sold_and_returned['sold_qty']
sold_and_returned[sold_and_returned['return_rate'] > 0].sort_values('return_rate', ascending=False)
```

```python
rfm = customers.groupby('CustomerID').agg(
    frequency = ('InvoiceNo', 'nunique'),
    monetary = ('TotalPrice', 'sum')
)
rfm['recency'] = (df['InvoiceDate'].max() - LastBuy).dt.days
rfm.head(5)
```

compete every customers' rfm score

```python
rfm['M'] = pd.qcut(rfm['monetary'], 5, labels=[1,2,3,4,5]).astype(int)
rfm['R'] = pd.qcut(rfm['recency'], 5, labels=[5,4,3,2,1]).astype(int)
rfm['F'] = pd.qcut(rfm['frequency'].rank(method='first'), 5, labels=[1,2,3,4,5]).astype(int)
rfm['RFM_score'] = rfm['M']+rfm['R']+rfm['F']
```
